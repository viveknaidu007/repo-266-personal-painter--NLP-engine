```python
import google.generativeai as genai
from dotenv import load_dotenv
import os

load_dotenv()

YOUR_API_KEY = os.getenv(GEMINI_API_KEY)
# Replace with your actual API key
API_KEY = "YOUR_API_KEY"

# Configure the Gemini API
genai.configure(api_key=API_KEY)

# Load the Gemini Pro model
model = genai.GenerativeModel('gemini-pro')

try:
    # Send a simple text generation request
    response = model.generate_content("Write a very short poem about a flower.")
    response.resolve() # This will raise an exception if there's an error

    # If the request was successful, print the generated text
    print("API key is working!")
    print("Generated text:")
    print(response.text)

except Exception as e:
    print(f"Error testing API key: {e}")
    print("Please double-check your API key and network connection.")
```
